## 1. Setup

## 2. Random Integers and Floating-Point Numbers

NumPy's `random` module (via the modern `Generator` API, `np.random.default_rng`) can produce
random integers and random floats drawn from a uniform distribution.

```python
# Create a random number generator
rng = np.random.default_rng()

# --- Random integers ---
single_int = rng.integers(1, 100)              # single random integer between 1 and 99
int_array = rng.integers(1, 100, size=10)       # array of 10 random integers
int_matrix = rng.integers(0, 10, size=(3, 4))   # 3x4 matrix of random integers 0-9

print("Single random integer (1-99):", single_int)
print("Array of 10 random integers (1-99):", int_array)
print("3x4 matrix of random integers (0-9):\n", int_matrix)
```

```python
# --- Random floating-point numbers ---
single_float = rng.random()               # single float in [0.0, 1.0)
float_array = rng.random(10)              # array of 10 floats in [0.0, 1.0)
uniform_range = rng.uniform(5, 15, size=5)  # 5 floats uniformly distributed between 5 and 15
normal_dist = rng.normal(loc=0, scale=1, size=5)  # 5 floats from a normal distribution

print("Single random float [0,1):", single_float)
print("Array of 10 random floats [0,1):", float_array)
print("5 floats uniform in [5, 15):", uniform_range)
print("5 floats from normal distribution (mean=0, std=1):", normal_dist)
```

## 3. Random Samples from an Array

`Generator.choice()` lets us draw random samples (with or without replacement) from an
existing array, and can also apply sampling weights/probabilities.

```python
fruits = np.array(["apple", "banana", "cherry", "date", "elderberry", "fig", "grape"])

# Sample WITHOUT replacement (each element can be picked only once)
sample_no_replace = rng.choice(fruits, size=3, replace=False)

# Sample WITH replacement (an element can be picked multiple times)
sample_with_replace = rng.choice(fruits, size=5, replace=True)

# Weighted sampling: 'grape' is 5x more likely to be picked than the others
weights = np.array([1, 1, 1, 1, 1, 1, 5])
weights = weights / weights.sum()
weighted_sample = rng.choice(fruits, size=5, replace=True, p=weights)

print("Original array:", fruits)
print("Sample without replacement (3 items):", sample_no_replace)
print("Sample with replacement (5 items):", sample_with_replace)
print("Weighted sample (grape favored):", weighted_sample)
```

```python
# Shuffling an array in place is another common sampling-related operation
deck = np.arange(1, 11)
print("Before shuffle:", deck)
rng.shuffle(deck)
print("After shuffle: ", deck)
```

## 4. Reproducible Results Using a Seed

By default, every run of a random function gives different results. Setting a **seed**
makes the sequence of "random" numbers deterministic and repeatable — essential for
debugging and for sharing reproducible experiments with others.

```python
# Without a seed: results differ every time the cell is run
gen_a = np.random.default_rng()
gen_b = np.random.default_rng()
print("Generator A:", gen_a.integers(0, 100, size=5))
print("Generator B:", gen_b.integers(0, 100, size=5))
print("Are they equal?", np.array_equal(gen_a.integers(0, 100, size=5), gen_b.integers(0, 100, size=5)))
```

```python
# With a seed: results are identical every single run
seeded_gen_1 = np.random.default_rng(seed=42)
seeded_gen_2 = np.random.default_rng(seed=42)

result_1 = seeded_gen_1.integers(0, 100, size=5)
result_2 = seeded_gen_2.integers(0, 100, size=5)

print("Seeded Generator 1:", result_1)
print("Seeded Generator 2:", result_2)
print("Are they equal?", np.array_equal(result_1, result_2))
```

## 5. Generating a Small Simulated Dataset

Putting it all together: a small reproducible synthetic dataset of "student marks",
similar in spirit to the suggested **Student Marks Dataset**.

```python
rng_seeded = np.random.default_rng(seed=7)

n_students = 10
student_ids = np.arange(1, n_students + 1)
# Marks simulated from a normal distribution, clipped to a valid 0-100 range
marks = rng_seeded.normal(loc=70, scale=12, size=n_students)
marks = np.clip(marks, 0, 100).round(1)

print("Student ID :", student_ids)
print("Marks      :", marks)
print("Average mark:", marks.mean().round(2))
```

```python
# Verify reproducibility of the simulated dataset: re-running with the same seed
# produces the exact same 'random' marks.
rng_check = np.random.default_rng(seed=7)
marks_check = np.clip(rng_check.normal(loc=70, scale=12, size=n_students), 0, 100).round(1)

print("Original marks: ", marks)
print("Recomputed marks:", marks_check)
print("Identical?", np.array_equal(marks, marks_check))
```

## 6. Interview Questions

**Q1. Why are random numbers useful in data science?**

Random numbers are used to simulate data when real data isn't available, to split
datasets into training/validation/test sets, to initialize model parameters (e.g. neural
network weights), to perform bootstrapping and cross-validation, to run Monte Carlo
simulations, and to shuffle data so that models don't learn spurious patterns tied to
the original data order.

**Q2. What is a random seed?**

A random seed is a starting value given to a pseudo-random number generator. Computers
can't generate *truly* random numbers — they use deterministic algorithms that produce
a long sequence of numbers that only *look* random. The seed determines exactly where in
that sequence the generator starts, so the same seed always produces the same sequence
of "random" numbers.

**Q3. Why is reproducibility important?**

Reproducibility means that an experiment, when re-run under the same conditions,
produces the same results. This matters because it lets other people (or your future
self) verify results, debug code reliably, fairly compare different models or
approaches on identical data splits/initializations, and trust that reported results
aren't just a lucky roll of the dice.
